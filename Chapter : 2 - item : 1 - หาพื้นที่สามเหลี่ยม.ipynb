```python
try:
    user_input = input("Enter number : ")

    b=int(user_input.split(",")[0].strip())
    h=int(user_input.split(",")[1].strip())

    # area = ½ x ฐาน x สูง
    # Calculate the area
    area = (b * h) / 2

    print(int(area))

except:
    print("error")
```
